```python
df['overview']=df['overview'].fillna('')
```

```python
import ast

def parse_genres(genres_str):
    if isinstance(genres_str, str):
        try:
            genres_list = ast.literal_eval(genres_str)
            if isinstance(genres_list, list):
                return ', '.join([d['name'] for d in genres_list if 'name' in d])
        except (ValueError, SyntaxError):
            pass
    return ''
```

```python
df['genres'] = df['genres'].apply(parse_genres)
```

```python
df=df.dropna(subset='title')
```

```python
df
```

```python
df['popularity']=df['popularity'].astype(float)
```

```python
df['tagline']=df['tagline'].fillna('')
df['vote_average']=df['vote_average'].fillna(df['vote_average'].mean())
df['popularity']=df['popularity'].fillna(df['popularity'].mean())
```

```python
df.isnull().sum()
```

```python
df['tags']=df['overview']+df['genres']+df['tagline']
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('wordnet')
nltk.download('stopwords')
```

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
def text_preprocessing(text):
  text=str(text).lower()
  text=re.sub(r'[^\w\s]', ' ', text)
  words=text.split()
  words=[word for word in words if word not in stop_words]
  words=[lemmatizer.lemmatize(word) for word in words]
  return ' '.join(words)
```

```python
df['tags']=df['tags'].apply(text_preprocessing)
```

```python
df
```

```python
df=df.reset_index(drop=True)
```

```python
indices=pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf_idf=TfidfVectorizer(stop_words='english',ngram_range=(1,2))
```

```python
tf_idf_matrix=tf_idf.fit_transform(df['tags'])
```

```python
print(tf_idf.get_feature_names_out())
print(tf_idf_matrix.toarray())
```

```python
tf_idf_matrix.shape
```

```python
from sklearn.metrics.pairwise import linear_kernel
```

```python
def recommend(title,n=10):
  if title not in indices:
    return('Movie not found')
  idx=indices[title]
  sim_score=linear_kernel(tf_idf_matrix[idx],tf_idf_matrix).flatten()
  similar_idx=sim_score.argsort()[::-1][1:n+1]
  return(df['title']).iloc[similar_idx]
```

```python
recommend('The Ouija Exorcism')
```
